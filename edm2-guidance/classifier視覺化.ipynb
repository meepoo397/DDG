```python
import os
from PIL import Image
import matplotlib.pyplot as plt

def show_images_from_dir(folder_path, num_images=4, start_seed=42):
    """
    顯示指定資料夾中的圖片（根據檔名排序，從指定種子碼開始），以 2x2 格顯示。

    Args:
        folder_path (str): 圖片所在資料夾路徑
        num_images (int): 要顯示的圖片數量（預設 4）
        start_seed (int): 從第幾個種子（圖片）開始（根據檔名排序）
    """
    if not os.path.isdir(folder_path):
        print(f"[錯誤] 資料夾不存在：{folder_path}")
        return

    # 過濾圖片檔案，並根據檔名中的數字排序
    image_files = sorted([
        os.path.join(folder_path, f)
        for f in os.listdir(folder_path)
        if f.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp'))
    ], key=lambda x: int(''.join(filter(str.isdigit, os.path.basename(x))) or 0))

    # 找到起始 index
    start_index = next((i for i, f in enumerate(image_files)
                        if int(''.join(filter(str.isdigit, os.path.basename(f))) or 0) >= start_seed), None)

    if start_index is None:
        print(f"[警告] 找不到 seed >= {start_seed} 的圖片。")
        return

    # subset = image_files[start_index:start_index + num_images]
    # if not subset:
    #     print(f"[警告] 從 seed {start_seed} 起沒有足夠圖片可顯示。")
    #     return
    img_path = image_files[start_index]
    # 顯示 2x2 圖片
    rows = cols = 1
    plt.figure(figsize=(8, 8))
    img = Image.open(img_path)
    plt.imshow(img)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def shift_and_scale_curve(data: list, min_guidance: float, max_guidance: float,
                          num_steps: int, sigma_min: float, sigma_max: float, rho: float,
                          interval_sigma_min: float = 0.28, interval_sigma_max: float = 2.9,
                          set_to_value: float = 1.0) -> np.ndarray:
    """
    Shifts and scales a 1D curve based on a specific t_steps condition.
    Values outside the condition range are set to a specified value.

    Parameters:
    - data (list): A list of numerical data representing the curve.
    - min_guidance (float): The desired minimum value for the new curve.
    - max_guidance (float): The desired maximum value for the new curve.
    - num_steps (int): The total number of steps for t_steps calculation.
    - sigma_min (float): The minimum sigma value.
    - sigma_max (float): The maximum sigma value.
    - rho (float): The rho parameter.
    - interval_sigma_min (float): The minimum t_step value of the condition interval.
    - interval_sigma_max (float): The maximum t_step value of the condition interval.
    - set_to_value (float): The value to set for data points outside the interval.

    Returns:
    - np.ndarray: A new NumPy array with the transformed data.
    """
    if not data:
        return np.array([])
    
    data = np.array(data)
    
    # 步驟1: 根據你的公式計算每個步驟的 t_steps
    step_indices = np.arange(num_steps)
    t_steps = (sigma_max**(1/rho) + step_indices / (num_steps - 1) * (sigma_min**(1/rho) - sigma_max**(1/rho)))**rho
    
    # 為了與你提供的數據長度匹配，我們只考慮前 num_steps 個 t_steps
    if len(t_steps) > len(data):
        t_steps = t_steps[:len(data)]
    
    # 步驟2: 根據 t_steps 判斷每個數據點是否符合條件
    is_guided = (t_steps >= interval_sigma_min) & (t_steps <= interval_sigma_max)
    
    # 步驟3: 提取符合條件的數據點
    guided_data = data[is_guided]
    
    # 創建一個新的陣列，預設所有值為 set_to_value
    new_data = np.full_like(data, set_to_value, dtype=float)

    if guided_data.size > 0:
        min_val = np.min(guided_data)
        max_val = np.max(guided_data)
        
        # 避免除以零
        if not np.isclose(min_val, max_val):
            # 對符合條件的數據進行 Rescale
            scaled_guided_data = (guided_data - min_val) / (max_val - min_val)
            transformed_guided_data = scaled_guided_data * (max_guidance - min_guidance) + min_guidance
            
            # 將轉換後的數據放回新的陣列中
            new_data[is_guided] = transformed_guided_data
        else:
            # 如果符合條件的數據範圍為零，則全部設定為 min_guidance
            new_data[is_guided] = min_guidance
    
    return new_data

# 原始數據 (與你提供的一致)
original_data = [0, 0.00011854968033730984, -6.204703822731972e-06, 7.344386540353298e-05, 8.147815242409706e-05, 7.347413338720798e-05, 0.00013053789734840393, 0.00024396670050919056, 0.00034406152553856373, 0.0004322801250964403, 0.0007102922536432743, 0.0017681904137134552, 0.00401817774400115, 0.00889554899185896, 0.019199594855308533, 0.033939484506845474, 0.06823297590017319, 0.1106177419424057, 0.12362787127494812, 0.11659982800483704, 0.09012675285339355, 0.053175389766693115, 0.02519327402114868, 0.008260965347290039, 0.0005714893341064453, -0.0017622709274291992, -0.0018764138221740723, -0.0013113021850585938, -0.0007520318031311035, -0.00029480457305908203, -0.0004126429557800293, -7.152557373046875e-07]

# 參數設定
min_g = 1.75
max_g = 2.8
num_steps = 32
sigma_min = 0.002
sigma_max = 80.0
rho = 7.0

# 計算新的曲線
new_data_curve = shift_and_scale_curve(original_data, min_g, max_g,
                                       num_steps, sigma_min, sigma_max, rho)

# 繪製曲線
plt.figure(figsize=(12, 8))

# 原始曲線
plt.plot(original_data, label=f'Original\n(min={np.min(original_data):.3f}, max={np.max(original_data):.3f})', color='blue')

# 轉換後的曲線
plt.plot(new_data_curve, label=f'Scaled (t in [{0.28}, {2.9}])', color='red', linestyle='--')

# 繪製參考線
plt.axhline(y=min_g, color='red', linestyle=':', linewidth=1, label=f'min_guidance ({min_g})')
plt.axhline(y=max_g, color='red', linestyle=':', linewidth=1, label=f'max_guidance ({max_g})')

# 繪製區間範圍
t_steps = (sigma_max**(1/rho) + np.arange(num_steps) / (num_steps - 1) * (sigma_min**(1/rho) - sigma_max**(1/rho)))**rho
filtered_t_steps_indices = np.where((t_steps >= 0.28) & (t_steps <= 2.9))[0]
if filtered_t_steps_indices.size > 0:
    plt.axvspan(xmin=filtered_t_steps_indices[0], xmax=filtered_t_steps_indices[-1],
                color='green', alpha=0.1, label=f'Rescale Interval (t_steps)')

plt.title('W Scheduled by Rescaling Difference of True Prob. Curve', fontsize=16)
plt.xlabel('Denoise Steps', fontsize=12)
plt.ylabel('Guidance Scale w', fontsize=12)
plt.legend()
plt.grid(True)
plt.show()
```

```python
def plot_prob_ratios(all_classifier_results_tensor, thresholds):
    """
    計算並繪製 P_true / P_pred 大於指定門檻值的比例。

    Args:
        all_classifier_results_tensor (torch.Tensor): 
            形狀為 (B, T, 7) 的 tensor，包含分類器結果。
        thresholds (list or tuple): 
            一個包含待測門檻值的列表，例如 [0.9, 0.8, 0.7]。
    """
    T = all_classifier_results_tensor.shape[1]
    
    # 創建一個字典來儲存每個門檻值的比例
    # 例如：{'2': [], '1': [], '0.5': []}
    prob_ratio_proportions = {str(t): [] for t in thresholds}

    # 迴圈遍歷每個時間 t
    for t in range(T):
        # 提取當前時間步 t 的預測機率和真實機率
        # 修正: pred_probs 在 channel 2，true_probs 在 channel 6
        pred_probs = all_classifier_results_tensor[:, t, 2]
        true_probs = all_classifier_results_tensor[:, t, 6]
        
        # 計算機率比值，並處理除以零的情況
        # torch.div 會在除以零時返回 inf
        prob_ratios = torch.div(true_probs, pred_probs)
        
        # 針對每個門檻值，計算滿足條件的比例
        for threshold in thresholds:
            # 判斷比值是否大於門檻值
            # (prob_ratios > threshold) 會得到一個布林 tensor
            # .float().mean().item() 計算比例
            ratio_proportion = (prob_ratios > threshold).float().mean().item()
            prob_ratio_proportions[str(threshold)].append(ratio_proportion)

    # --- 繪製圖表 ---
    plt.figure(figsize=(10, 6))
    for threshold, proportions in prob_ratio_proportions.items():
        plt.plot(range(T), proportions, marker='o', linestyle='-', label=f'P_true / P_pred > {threshold}')
        
    plt.title('Proportion of Samples Meeting Specific Probability Ratio Criteria')
    plt.xlabel('Time Step (t)')
    plt.ylabel('Proportion')
    plt.legend()
    plt.grid(True)
    plt.show()
test_thresholds = [0.9,0.8,0.7,0.6,0.5,0.4,0.3,0.2]

# 呼叫函式來繪製圖表
plot_prob_ratios(all_classifier_results_tensor, test_thresholds)
```

```python
print(f"Loaded tensor shape: {all_classifier_results_tensor.shape}")

# 獲取時間步長 T
T = all_classifier_results_tensor.shape[1]

# 創建列表來儲存每個時間步的平均比值
average_prob_ratios = []

# 迴圈遍歷每個時間 t
for t in range(T):
    # 提取當前時間步 t 的資料
    # 預測標籤：索引 1
    pred_labels = all_classifier_results_tensor[:, t, 1]
    # 第二高預測機率：索引 4
    pred_probs_2nd = all_classifier_results_tensor[:, t, 4]
    # 真實標籤：索引 5
    true_labels = all_classifier_results_tensor[:, t, 5]
    # 真實標籤的預測機率：索引 6
    true_probs = all_classifier_results_tensor[:, t, 6]
    
    # 計算並篩選出預測正確的樣本
    correct_mask = (pred_labels == true_labels)
    correct_true_probs = true_probs[correct_mask]
    correct_pred_probs_2nd = pred_probs_2nd[correct_mask]
    
    # 如果沒有預測正確的樣本，則該時間步的平均比值為 0
    if correct_true_probs.numel() == 0:
        average_prob_ratios.append(0.0)
        continue
    
    # 計算比值並取平均
    # 處理除以 0 的情況，並計算平均值
    prob_ratios = torch.div(correct_true_probs, correct_pred_probs_2nd)
    avg_ratio = prob_ratios.mean().item()
    average_prob_ratios.append(avg_ratio)

# --- 繪製圖表 ---

plt.figure(figsize=(10, 6))
plt.plot(range(T), average_prob_ratios, marker='o', linestyle='-')
plt.title('Average Ratio of P_true / P_pred_2 for Correct Predictions')
plt.xlabel('Time Step (t)')
plt.ylabel('Average Ratio')
plt.grid(True)
plt.show()
```
