```python
import pickle
import matplotlib.pyplot as plt
import os

# 設定目錄路徑
# 請確保你的 .pkl 檔案都在這個目錄下
dir_path = "."

# 創建一個新的圖表
plt.figure(figsize=(10, 6))

# 遍歷目錄中的所有檔案
for filename in os.listdir(dir_path):
    # 檢查檔案是否以 .pkl 結尾
    if filename.endswith(".pkl"):
        file_path = os.path.join(dir_path, filename)

        # 讀取 .pkl 檔案
        with open(file_path, 'rb') as f:
            accuracy_list_loaded = pickle.load(f)

        # 提取檔案名稱的前綴作為標籤
        # 例如: "res50v2_nofinetune.pkl" -> "res50v2_nofinetune"
        label = os.path.splitext(filename)[0]

        # 假設你的 x 軸數據是從 0 開始，以 2 為步長
        x_values = [i * 2 for i in range(len(accuracy_list_loaded))]

        # 繪製曲線並設定標籤
        plt.plot(x_values, accuracy_list_loaded, label=label)

# 添加標題和軸標籤
plt.title("Accuracy of x0 estimation")
plt.xlabel("Diffusion steps")
plt.ylabel("Accuracy")

# 顯示圖例
plt.legend()

# 顯示圖表
plt.grid(True)
plt.show()
```
